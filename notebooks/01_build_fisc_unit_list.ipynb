```python
import os
import sys
import pandas as pd
import numpy as np
import dotenv

import warnings
warnings.filterwarnings("ignore", category=pd.errors.PerformanceWarning)

dotenv.load_dotenv(dotenv.find_dotenv())

ROOT_PATH = os.getenv("ROOT_PATH")
MY_DATA_PATH = os.getenv("MY_DATA_PATH")
RAW_DATA_PATH = os.getenv("RAW_DATA_PATH")

sys.path.append(os.path.join(ROOT_PATH, "scripts"))

INPUT_FILEPATH = os.path.join(RAW_DATA_PATH, "lincoln-institute/FiSC-Full-Dataset-2023-Update.xlsx")
OUTPUT_FILEPATH = os.path.join(MY_DATA_PATH, "raw_data/fisc_unit_list.csv")
```
