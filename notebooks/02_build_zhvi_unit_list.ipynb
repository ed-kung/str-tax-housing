```python
import os
import sys
import pandas as pd
import numpy as np
import dotenv

import warnings
warnings.filterwarnings("ignore", category=pd.errors.PerformanceWarning)

dotenv.load_dotenv(dotenv.find_dotenv())

ROOT_PATH = os.getenv("ROOT_PATH")
MY_DATA_PATH = os.getenv("MY_DATA_PATH")
RAW_DATA_PATH = os.getenv("RAW_DATA_PATH")

sys.path.append(os.path.join(ROOT_PATH, "scripts"))

INPUT_FILEPATH = os.path.join(RAW_DATA_PATH, "zhvi/City_zhvi_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.csv")
OUTPUT_FILEPATH = os.path.join(MY_DATA_PATH, "raw_data/zhvi_unit_list.csv")
```

```python
df = pd.read_csv(INPUT_FILEPATH)

df = (
    df[['RegionID', 'RegionName', 'StateName', 'RegionType']]
    .rename(columns={
        'RegionID': 'zhvi_id', 
        'RegionName': 'region_name',
        'StateName': 'state_name',
        'RegionType': 'region_type'
    })
    .drop_duplicates()
    .reset_index(drop=True)
)

df.to_csv(OUTPUT_FILEPATH, index=False)
df.info()
```
